```python
segment_profit=df.groupby(['Segment','Country'])['Profit'].sum()
segment_profit=segment_profit.sort_values(ascending=False)
print(segment_profit)
plt.yticks(np.arange(segment_profit.min(),segment_profit.max()+1,10000)*10)
segment_profit.plot(kind='bar')
plt.xlabel('segment and country')
plt.ylabel('profit in lakhs')
plt.title('segment and country wise profits')
plt.show()
```

```python
segment_profit=df.groupby(['Segment','Product'])['Profit'].sum()
segment_profit=segment_profit.sort_values(ascending=False)
print(segment_profit)
plt.yticks(np.arange(segment_profit.min(),segment_profit.max()+1,10000)*10)
segment_profit.plot(kind='bar')
plt.xlabel('segment and country')
plt.ylabel('profit in lakhs')
plt.title('segment and product wise profits')
plt.show()
```

```python
import pandas as pd
import matplotlib.pyplot as plt
import numpy as np

# Create pivot table
segment_sale = df.pivot_table(
    index='Segment',
    columns='Product',
    values=' Sales',
    aggfunc='sum',
    fill_value=0
)

# Display pivot table
print("Segment and Product Wise Sales:")
display(segment_sale)

# Convert Sales into Lakhs
segment_sale_lakhs = segment_sale / 100000

"""# Display Sales in Lakhs
print("Sales in Lakhs:")
display(segment_sale_lakhs)"""

# Plot
segment_sale_lakhs.plot(
    kind='bar',
    figsize=(12, 6)
)

# Y-axis intervals
plt.yticks(
    np.arange(
        0,
        segment_sale_lakhs.max().max() + 1,
        1
    )
)

plt.xlabel('Segment')
plt.ylabel('Sales (Lakhs)')
plt.title('Segment and Product Wise Sales')
plt.xticks(rotation=45)
plt.legend(title='Product')
plt.tight_layout()
plt.show()
```
